```python
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

import string
```

```python
text = """The movie was entertaining from beginning to end and featured excellent performances by the lead actors. The visual effects and background music added to the overall experience, making several scenes memorable. Although the story slowed down slightly in the second half, the emotional ending made up for it. I would recommend this movie to anyone who enjoys action, drama, and inspiring stories."""

print(text)
```

```python
print("Task 1: Identify Text Type")
print("Text Type: Movie Review")
```

```python
sentences = sent_tokenize(text)

print("Task 2: Sentence Segmentation\n")

for i, sentence in enumerate(sentences, 1):
    print(f"Sentence {i}: {sentence}\n")
```

```python
tokens = word_tokenize(text)

print("Task 3: Word Tokenization\n")
print(tokens)
```

```python
tokens = [word.lower() for word in tokens]

tokens = [word for word in tokens if word.isalpha()]

print("Task 4: Lowercase and Remove Punctuation & Numbers\n")
print(tokens)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word not in stop_words]

print("Task 5: Remove Stop Words\n")
print(filtered_tokens)
```

```python
stemmer = PorterStemmer()

stemmed_tokens = [stemmer.stem(word) for word in filtered_tokens]

print("Task 6: Stemming\n")
print(stemmed_tokens)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]

print("Task 7: Lemmatization\n")
print(lemmatized_tokens)
```

```python
count_vectorizer = CountVectorizer()

count_matrix = count_vectorizer.fit_transform([" ".join(lemmatized_tokens)])

print("Task 8: Count Vectorization\n")
print("Features:", count_vectorizer.get_feature_names_out())
print("\nVector:")
print(count_matrix.toarray())
```

```python
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform([" ".join(lemmatized_tokens)])

print("Task 9: TF-IDF Vectorization\n")
print("Features:", tfidf_vectorizer.get_feature_names_out())
print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

```python
positive_words = [
    "excellent",
    "entertaining",
    "memorable",
    "recommend",
    "emotional",
    "inspiring"
]

negative_words = [
    "slowed"
]

print("Task 10: Sentiment Analysis\n")
print("Positive Words:", positive_words)
print("Negative Words:", negative_words)
print("\nOverall Sentiment: POSITIVE")
print("\nReason: The review contains many positive words and only one minor negative expression, so the overall sentiment is Positive.")
```
